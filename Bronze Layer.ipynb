```python
from pyspark.sql.functions import current_timestamp, col

df_bronze = spark.read.csv(
    "/Workspace/Users/anidesifitriaaaa@gmail.com/Drafts/healthcare_api_data",
    header=True,
    inferSchema=True).withColumn(
    "ingesttime", current_timestamp()).withColumn(
    "filename", col("_metadata.file_path") )

display(df_bronze)
```

```python
# Remove columns with spaces, parentheses, and dashes
for col_name in df_bronze.columns:
    clean_name = col_name.strip() \
        .replace(" ", "_") \
        .replace("(", "") \
        .replace(")", "") \
        .replace("-", "_")
   
    df_bronze = df_bronze.withColumnRenamed(col_name, clean_name)
```

```python
bronze_save = df_bronze.write.mode("overwrite").format("delta").saveAsTable("Bronze_Data_Table")
print("succesfully saved!")
```
